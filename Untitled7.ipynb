```python
import numpy as np


def build_board_arrays(population, output_alphabet):
    pop_size = len(population)
    n_deflectors = len(population[0].deflectors_list)

    dest = np.empty((pop_size, n_deflectors), dtype=np.int64)
    is_hole = np.empty((pop_size, n_deflectors), dtype=bool)
    bounce_val = np.zeros((pop_size, n_deflectors), dtype=np.int64)
    emit_code = np.full((pop_size, n_deflectors), -1, dtype=np.int64)

    out_index = {ch: i for i, ch in enumerate(output_alphabet)}

    for p, brain in enumerate(population):
        for d, (dst, val) in enumerate(brain.deflectors_list):
            dest[p, d] = dst
            if isinstance(val, int):
                bounce_val[p, d] = val
            else:
                is_hole[p, d] = True
                emit_code[p, d] = out_index[val]

    directions = np.array([b.alphabet_directions for b in population], dtype=np.int64)
    start_pos = np.array([b.alphabet_start_positions for b in population], dtype=np.int64)

    return {
        "dest": dest, "is_hole": is_hole, "bounce_val": bounce_val,
        "emit_code": emit_code, "directions": directions, "start_pos": start_pos,
        "n_deflectors": n_deflectors, "output_alphabet": output_alphabet,
    }


def vectorized_compute(board, tokens, ball_energy, cap):
    pop_size, n_deflectors = board["dest"].shape
    num_tokens = len(tokens)
    if num_tokens == 0:
        return [[] for _ in range(pop_size)]

    tok = np.array(tokens, dtype=np.int64)
    dirs = board["directions"][:, tok]            # (pop, tokens)
    starts = board["start_pos"][:, tok]            # (pop, tokens)
    total = dirs.sum(axis=1, keepdims=True)         # vectorized `travel`:
    travel_start = starts + total - dirs            # sum(all)-self, no per-token O(n) pop/sum

    idx = travel_start % n_deflectors
    energy = np.full((pop_size, num_tokens), ball_energy, dtype=np.int64)
    active = np.ones((pop_size, num_tokens), dtype=bool)
    row = np.arange(pop_size)[:, None]

    emissions = np.full((cap, pop_size, num_tokens), -1, dtype=np.int64)

    for step in range(cap):
        dest_i = board["dest"][row, idx]
        hole_i = board["is_hole"][row, idx]
        bval_i = board["bounce_val"][row, idx]
        eval_i = board["emit_code"][row, idx]

        distance = np.abs(dest_i - idx)
        delta = np.where(hole_i, 1, bval_i)
        new_energy = np.where(active, energy + delta - distance, energy)
        new_idx = np.where(active, dest_i, idx)

        emitting = active & hole_i
        emissions[step] = np.where(emitting, eval_i, -1)

        energy = new_energy
        idx = new_idx
        active = active & (energy > 0)

    # emissions is (cap, pop, tokens); reorder to (pop, tokens, cap) so that,
    # per brain, flattening gives token-major/step-minor order -- matching
    # the original "for token: for step: maybe emit" ordering exactly.
    flat = emissions.transpose(1, 2, 0).reshape(pop_size, -1)
    out_alpha = np.array(board["output_alphabet"], dtype=object)

    results = []
    for row in flat:
        codes = row[row != -1]                 # one fast C-level filter per brain
        results.append(out_alpha[codes].tolist())
    return results
```

```python
import random
import math
import copy

def TokenizeInput(input, alphabet):
    input_str = str(input)
    matches = []

    for order, item in enumerate(alphabet):
        item_str = str(item)
        if not item_str:
            continue

        index = input_str.find(item_str)
        if index != -1:
            matches.append((index, -len(item_str), order, item))

    matches.sort(key=lambda match: (match[0], match[1], match[2]))
    return [item for _, _, _, item in matches]

class Ball_Trap_Brain:
    def __init__(self, deflectors, bounce, alphabet, output, holes_multiplier=1):
        self.output = output
        self.holes_multiplier = holes_multiplier
        self.ball_energy = bounce*2
        self.alphabet = alphabet
        self.alphabet_directions = self.choose_alphabet_direction(len(alphabet))
        self.alphabet_start_positions = self.start_positions(len(alphabet), deflectors)
        self.deflectors_list = [[random.randint(0, deflectors-1), bounce] for i in range(deflectors)]
        self.set_holes(output, deflectors, self.holes_multiplier)

    def set_holes (self, output, deflectors, holes_multiplier):
        for i in range(len(output)):
          for j in range(holes_multiplier):
            self.deflectors_list[random.randint(0, deflectors-1)][1] = output[i]

    def choose_alphabet_direction(self, alphabet):
        directions = [random.choice([1,-1]) for _ in range(alphabet)]
        return directions

    def start_positions(self, alphabet, deflectors):
        return [random.randint(0, deflectors-1) for i in range(alphabet)]


    def compute_input(self, input):
        directions = []
        positions = []
        travel = []
        input_arr = TokenizeInput(input, self.alphabet)

        for i in input_arr:
            if i in self.alphabet:
                index = self.alphabet.index(i)
                directions.append(self.alphabet_directions[index])
                positions.append(self.alphabet_start_positions[index])

        for i in range(len(directions)):
            travel.append(self.travel(copy.copy(directions), positions[i], i))


        return self.ricochet(travel, 1, self.ball_energy)

    def ricochet(self, travel, cap, ball_energy):
        output = []
        dl = self.deflectors_list
        n = len(dl)

        for t in travel:
            # normalize index using modulo (handles large or negative values)
            new_idx = t % n
            cap_idx = 0
            be = ball_energy

            while be > 0 and cap_idx < cap:
                dest, val = dl[new_idx]
                distance = abs(dest - new_idx)

                if not isinstance(val, int):
                    output.append(val)
                    be += 1
                    be -= distance
                    new_idx = dest
                    cap_idx += 1
                    continue

                be += val
                be -= distance
                new_idx = dest
                cap_idx += 1

        return output


    def travel(self, directions, position, index):
        directions.pop(index)
        position += sum(directions)
        return position

    def display_deflectors(self):
        for i, deflect in enumerate(self.deflectors_list):
            print(f"Deflector {i}: {deflect}")

    def clone(self):
      # Create a lightweight dummy instance
      new_brain = Ball_Trap_Brain(1, self.ball_energy // 2, self.alphabet, self.output, self.holes_multiplier)


      new_brain.deflectors_list = [[d, v] for d, v in self.deflectors_list]
      new_brain.alphabet_directions = self.alphabet_directions[:]
      new_brain.alphabet_start_positions = self.alphabet_start_positions[:]

      return new_brain

    def shift_alphabet(self):
        random.shuffle(self.alphabet)
        for i in range(len(self.deflectors_list)):
          if self.deflectors_list[i][1] in self.output:
            idx = self.output.index(self.deflectors_list[i][1])
            self.deflectors_list[i][1] = self.alphabet[idx]

    def Mutate(self, individual_probability=0.1, probability=0.05):
        if random.random() < individual_probability:
            for i in range(len(self.deflectors_list)):
                  if random.random() < probability:
                      self.deflectors_list[i][0] = random.randint(0, len(self.deflectors_list)-1)

            for i in range(len(self.deflectors_list)):
                 if random.random() < probability:
                     self.deflectors_list[random.randint(0, len(self.deflectors_list)-1)][1] = random.choice(self.output)

            for i in range(len(self.alphabet_directions)):
                 if random.random() < probability:
                     self.alphabet_directions[i] = 1 if self.alphabet_directions[i] == -1 else -1

            for i in range(len(self.alphabet_start_positions)):
                  if random.random() < probability:
                      self.alphabet_start_positions[i] = random.randint(0, len(self.deflectors_list)-1)

            if random.random() < probability:
                  self.set_holes(self.output, len(self.deflectors_list), self.holes_multiplier)

        return self
```

```python
import random
import time
import sys


random.seed(0)

alphabet = [f'{i}' for i in range(10)] + ['+']
output = [f'{i}' for i in range(10)]

POP = 300
N_DEFLECTORS = 40
population = [Ball_Trap_Brain(N_DEFLECTORS, 3, alphabet, output, holes_multiplier=3) for _ in range(POP)]

test_inputs = ["1+9", "23+7", "5+5", "100+1", "0+0", "9+9+9"]


board = build_board_arrays(population, output)

all_match = True
for s in test_inputs:
    input_arr = TokenizeInput(s, alphabet)
    tokens = [alphabet.index(c) for c in input_arr if c in alphabet]

    vec_results = vectorized_compute(board, tokens, ball_energy=population[0].ball_energy, cap=1)

    for p, brain in enumerate(population):
        orig = brain.compute_input(s)
        vec = vec_results[p]
        if orig != vec:
            all_match = False
            print(f"MISMATCH input={s!r} brain={p}: orig={orig} vec={vec}")

print("All outputs match original implementation:" , all_match)


all_match_cap = True
for s in test_inputs:
    input_arr = TokenizeInput(s, alphabet)
    tokens = [alphabet.index(c) for c in input_arr if c in alphabet]
    vec_results = vectorized_compute(board, tokens, ball_energy=population[0].ball_energy, cap=5)
    for p, brain in enumerate(population):

        directions = []
        positions = []
        for i in input_arr:
            if i in brain.alphabet:
                idx = brain.alphabet.index(i)
                directions.append(brain.alphabet_directions[idx])
                positions.append(brain.alphabet_start_positions[idx])
        travel = [brain.travel(list(directions), positions[i], i) for i in range(len(directions))]
        orig = brain.ricochet(travel, 5, brain.ball_energy)
        vec = vec_results[p]
        if orig != vec:
            all_match_cap = False
            print(f"CAP5 MISMATCH input={s!r} brain={p}: orig={orig} vec={vec}")

print("All outputs match with cap=5:", all_match_cap)


big_pop_size = 1000
big_population = [Ball_Trap_Brain(100, 3, alphabet, output, holes_multiplier=3) for _ in range(big_pop_size)]
test_str = "23+789"
input_arr = TokenizeInput(test_str, alphabet)
tokens = [alphabet.index(c) for c in input_arr if c in alphabet]

t0 = time.perf_counter()
for brain in big_population:
    brain.compute_input(test_str)
t1 = time.perf_counter()
print(f"Original loop, pop={big_pop_size}: {t1 - t0:.4f}s")

t0 = time.perf_counter()
big_board = build_board_arrays(big_population, output)
t1 = time.perf_counter()
print(f"  (board-build overhead: {t1-t0:.4f}s  one-time per generation, not per test case)")

t0 = time.perf_counter()
vec_results = vectorized_compute(big_board, tokens, ball_energy=big_population[0].ball_energy, cap=1)
t1 = time.perf_counter()
print(f"Vectorized, pop={big_pop_size}: {t1 - t0:.4f}s")

mismatches = 0
for p in range(0, big_pop_size, 137):
    orig = big_population[p].compute_input(test_str)
    if orig != vec_results[p]:
        mismatches += 1
print("spot-check mismatches:", mismatches)


test_cases = [f"{i}+{j}" for i in range(20) for j in range(20)]  # 400 cases

t0 = time.perf_counter()
for brain in big_population:
    for s in test_cases:
        brain.compute_input(s)
t1 = time.perf_counter()
orig_total = t1 - t0
print(f"Original: pop={big_pop_size} x {len(test_cases)} cases = {orig_total:.3f}s")

t0 = time.perf_counter()
big_board = build_board_arrays(big_population, output)
for s in test_cases:
    input_arr = TokenizeInput(s, alphabet)
    tokens = [alphabet.index(c) for c in input_arr if c in alphabet]
    vectorized_compute(big_board, tokens, ball_energy=big_population[0].ball_energy, cap=1)
t1 = time.perf_counter()
vec_total = t1 - t0
print(f"Vectorized: pop={big_pop_size} x {len(test_cases)} cases = {vec_total:.3f}s")
print(f"Speedup: {orig_total / vec_total:.1f}x")
```
